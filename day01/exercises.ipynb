# Day 1 Exercises: Matrix Multiplication & Benchmarking


```python
import numpy as np
import time
```

### 1. Implement Matrix Multiplication from scratch
Write a function to multiply two matrices using nested python loops.


```python
def matmul_scratch(A, B):
    # Ensure dimensions match
    assert len(A[0]) == len(B), "Inner dimensions must match!"
    rows_A = len(A)
    cols_A = len(A[0])
    cols_B = len(B[0])
    
    # Initialize output matrix with zeros
    result = [[0.0 for _ in range(cols_B)] for _ in range(rows_A)]
    
    # Perform multiplication
    for i in range(rows_A):
        for j in range(cols_B):
            for k in range(cols_A):
                result[i][j] += A[i][k] * B[k][j]
    return result

# Test case
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
print("Scratch result:", matmul_scratch(A, B))

```

### 2. Compare with NumPy
Benchmark the scratch function against NumPy's `@` operator on large matrices.


```python
size = 100
A_np = np.random.randn(size, size)
B_np = np.random.randn(size, size)

A_list = A_np.tolist()
B_list = B_np.tolist()

# Benchmark Scratch
start = time.time()
res_scratch = matmul_scratch(A_list, B_list)
scratch_time = time.time() - start

# Benchmark NumPy
start = time.time()
res_np = A_np @ B_np
np_time = time.time() - start

print(f"Scratch time: {scratch_time:.4f}s")
print(f"NumPy time: {np_time:.4f}s")
print(f"Speedup: {scratch_time / np_time:.1f}x")

```
